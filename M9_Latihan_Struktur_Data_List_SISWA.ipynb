```python
# Isi nama dan nomor absenmu (lihat daftar absen di Google Classroom)
nama  = "..."
absen = 0   # ganti 0 dengan nomor absenmu (1 sampai 25)

harga_dasar = [252.3, 254.1, 251.8, 255.6, 257.2, 256.0, 258.9, 260.4, 259.1, 261.7]

# Data unikmu: setiap harga dasar dikalikan (1 + absen/100), dibulatkan 2 desimal
harga = []
for x in harga_dasar:
    harga.append(round(x * (1 + absen / 100), 2))

print("Nama :", nama, "| Absen:", absen)
print(harga)
print("Tipe data:", type(harga))
```

## Langkah 3: Slicing (Mengambil Sebagian List)
Format slicing adalah `list[awal:akhir]`. Elemen pada posisi **awal** ikut diambil, sedangkan elemen pada posisi **akhir tidak** ikut diambil.

| Penulisan | Arti |
|---|---|
| `x[0:3]` | elemen indeks 0, 1, 2 |
| `x[:3]` | 3 elemen pertama |
| `x[-3:]` | 3 elemen terakhir |
| `x[1:-1]` | semua elemen kecuali yang pertama dan terakhir |

```python
print("3 hari pertama :", harga[:3])
print("3 hari terakhir:", harga[-3:])
```

## Langkah 5: Mengubah Isi List
List bersifat **mutable**, artinya isinya dapat diubah setelah dibuat.

| Perintah | Kegunaan |
|---|---|
| `x.append(nilai)` | menambah elemen di akhir |
| `x.insert(indeks, nilai)` | menyisipkan elemen pada posisi tertentu |
| `x.remove(nilai)` | menghapus elemen berdasarkan nilainya |
| `x.pop()` | menghapus dan mengambil elemen terakhir |

Sel berikut bekerja pada **salinan** list (`.copy()`), sehingga data asli tidak berubah.

```python
data_baru = harga.copy()
data_baru.append(round(harga[-1] + 1.3, 2))
print("Setelah append:", data_baru)
```
